**Handling Inconsistency**

```python
#standardise inconsistencies
#login devices
df['PreferredLoginDevice'] = df['PreferredLoginDevice'].replace({
   'Mobile Phone': 'Mobile',
   'Phone': 'Mobile'})

#payment modes
df['PreferredPaymentMode'] = df['PreferredPaymentMode'].replace({
   'CC': 'Credit Card',
   'COD': 'Cash on Delivery'})

#order categories
df['PreferedOrderCat'] = df['PreferedOrderCat'].replace({
   'Mobile Phone': 'Mobile'})
```

```python
#cross check inconsistency
categorical_cols = df.select_dtypes(exclude=['number']).columns

for col in categorical_cols:
    display(df[col].value_counts())
```

**Imputing Data**

```python
#imputing missing values 
num_cols_with_na = [
    'Tenure', 'WarehouseToHome', 'HourSpendOnApp', 
    'OrderAmountHikeFromlastYear', 'CouponUsed', 
    'OrderCount', 'DaySinceLastOrder'
]

#grouped imputation by logical categories
df['WarehouseToHome'] = df['WarehouseToHome'].fillna(
    df.groupby('CityTier')['WarehouseToHome'].transform('median'))

df['HourSpendOnApp'] = df['HourSpendOnApp'].fillna(
    df.groupby('PreferredLoginDevice')['HourSpendOnApp'].transform('median'))

df['Tenure'] = df['Tenure'].fillna(
    df.groupby('PreferedOrderCat')['Tenure'].transform('median'))

df['OrderCount'] = df['OrderCount'].fillna(
    df.groupby('PreferedOrderCat')['OrderCount'].transform('median'))

#fallback: fill any remaining missing values with overall column medians
for col in num_cols_with_na:
    df[col] = df[col].fillna(df[col].median())

#cross check 0 missing values remain
print('missing values:', df[num_cols_with_na].isnull().sum().sum())
```

```python
#cross check missing values
missing = df.isnull().sum()
missing[missing > 0]
```

**Handling Data Types**

```python
#check dtypes
df.dtypes
```

```python
#convert count columns back to integers
count_cols = ['Tenure', 'WarehouseToHome', 'HourSpendOnApp', 
              'OrderAmountHikeFromlastYear', 'CouponUsed', 
              'OrderCount', 'DaySinceLastOrder']
df[count_cols] = df[count_cols].astype(int)

#convert CustomerID to string
df['CustomerID'] = df['CustomerID'].astype(str)

#convert categorical variables to 'category' dtype
cat_cols = ['PreferredLoginDevice', 'PreferredPaymentMode', 'Gender', 
            'PreferedOrderCat', 'MaritalStatus', 'CityTier']
df[cat_cols] = df[cat_cols].astype('category')
```

```python
#cross check updated data types
df.info()
```

**Feature Engineering**

```python
#ustainability & returns proxies
df['Estimated_Returns'] = np.where(df['Complain'] == 1, 
                          np.ceil(df['OrderCount'] * 0.20), 
                          np.ceil(df['OrderCount'] * 0.05))
df['Estimated_CO2_kg'] = df['Estimated_Returns'] * 1.2
```

```python
#tenure buckets (high-risk window vs retained)
df['TenureBucket'] = pd.cut(df['Tenure'], bins=[-1, 3, 12, 60, 100], 
                     labels=['0-3 Mos (High Risk)', '4-12 Mos', '1-5 Yrs', '5+ Yrs'])

#warehouse distance category (logistics & sustainability risk)
df['DistanceCategory'] = pd.cut(df['WarehouseToHome'], bins=[-1, 10, 20, 50, 200], 
                         labels=['Close (0-10km)', 'Moderate (11-20km)', 'Far (21-50km)', 'Very Far (>50km)'])
```

```python
#cross check feature outputs and check for missing values
print(df[['Estimated_Returns', 'Estimated_CO2_kg', 'TenureBucket', 'DistanceCategory']].isnull().sum())
```
